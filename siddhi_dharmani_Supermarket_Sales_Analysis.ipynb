# Supermarket Sales Analysis
## Data Analytics Project

This Google Colab-ready notebook analyzes 500 supermarket transactions.

**Questions answered**
- Which product generates the highest sales?
- Which branch performs best?
- Which category sells the most?
- What is the most popular payment method?
- Do Members spend more per transaction than Normal customers?
- What is the average customer rating?

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
df = pd.read_csv("supermarket_sales.csv")
df.head()
```

```python
# Data quality checks
print("Dataset shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nData types:")
print(df.dtypes)
```

```python
# Validate Sales = Quantity × Unit Price
df["Calculated_Sales"] = (df["Quantity"] * df["Unit_Price"]).round(2)
df["Sales_Match"] = np.isclose(df["Calculated_Sales"], df["Sales"], atol=0.01)
print("Matching rows:", int(df["Sales_Match"].sum()), "/", len(df))
```

```python
# Descriptive statistics
df.describe(include="all").T
```

```python
# 1. Highest-selling product
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)
display(product_sales.to_frame("Total_Sales"))
print(f"Highest-selling product: {product_sales.idxmax()} — ₹{product_sales.max():,.2f}")
```

```python
# 2. Best-performing branch
branch_sales = df.groupby(["Branch","City"])["Sales"].sum().sort_values(ascending=False)
display(branch_sales.to_frame("Total_Sales"))
b = branch_sales.index[0]
print(f"Best branch: Branch {b[0]} ({b[1]}) — ₹{branch_sales.iloc[0]:,.2f}")
```

```python
# 3. Highest-selling category
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
display(category_sales.to_frame("Total_Sales"))
print(f"Highest-selling category: {category_sales.idxmax()} — ₹{category_sales.max():,.2f}")
```

```python
# 4. Most popular payment method
payment_counts = df["Payment_Method"].value_counts()
display(payment_counts.to_frame("Transaction_Count"))
print(f"Most popular payment method: {payment_counts.idxmax()} — {payment_counts.max()} transactions")
```

```python
# 5. Average transaction value by customer type
customer_avg = df.groupby("Customer_Type")["Sales"].mean()
display(customer_avg.to_frame("Average_Transaction_Value"))
print(f"Member average: ₹{customer_avg.get('Member', np.nan):,.2f}")
print(f"Normal average: ₹{customer_avg.get('Normal', np.nan):,.2f}")
```

```python
# 6. Average customer rating
print(f"Average customer rating: {df['Rating'].mean():.2f} / 5")
```

```python
# Visualization 1: Sales by Product
plt.figure(figsize=(10,5))
sns.barplot(data=df, x="Product", y="Sales", estimator=sum, errorbar=None)
plt.title("Total Sales by Product")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Visualization 2: Sales by Branch
branch_chart = df.groupby("Branch", as_index=False)["Sales"].sum()
sns.barplot(data=branch_chart, x="Branch", y="Sales", errorbar=None)
plt.title("Total Sales by Branch")
plt.tight_layout()
plt.show()
```

```python
# Visualization 3: Sales by Category
category_chart = df.groupby("Category", as_index=False)["Sales"].sum()
sns.barplot(data=category_chart, x="Category", y="Sales", errorbar=None)
plt.title("Total Sales by Category")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Visualization 4: Payment methods
sns.countplot(data=df, x="Payment_Method", order=df["Payment_Method"].value_counts().index)
plt.title("Payment Method Usage")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
# Visualization 5: Customer type
customer_chart = df.groupby("Customer_Type", as_index=False)["Sales"].mean()
sns.barplot(data=customer_chart, x="Customer_Type", y="Sales", errorbar=None)
plt.title("Average Transaction Value by Customer Type")
plt.tight_layout()
plt.show()
```

```python
# Visualization 6: Ratings
sns.histplot(data=df, x="Rating", bins=10, kde=True)
plt.title("Customer Rating Distribution")
plt.tight_layout()
plt.show()
```

```python
# Final business summary
print("BUSINESS INSIGHTS")
print("-"*60)
print(f"1. Highest-selling product: {product_sales.idxmax()} — ₹{product_sales.max():,.2f}")
print(f"2. Best branch: Branch {branch_sales.index[0][0]} ({branch_sales.index[0][1]}) — ₹{branch_sales.iloc[0]:,.2f}")
print(f"3. Highest-selling category: {category_sales.idxmax()} — ₹{category_sales.max():,.2f}")
print(f"4. Most popular payment method: {payment_counts.idxmax()} — {payment_counts.max()} transactions")
print(f"5. Member average: ₹{customer_avg.get('Member', np.nan):,.2f}")
print(f"   Normal average: ₹{customer_avg.get('Normal', np.nan):,.2f}")
print(f"6. Average rating: {df['Rating'].mean():.2f}/5")
```
